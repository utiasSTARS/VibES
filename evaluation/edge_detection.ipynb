```python
# code to make edge maps for each of the clean baselines images
import numpy as np
import cv2 as cv
from matplotlib import pyplot as plt
from glob import glob
from tqdm import tqdm
 
RAW_IMAGE_PATH = 'raw_images'
EDGE_IMAGE_PATH = 'edge_images'

images = glob(RAW_IMAGE_PATH + '/*.png')
for image in tqdm(images):
    img = cv.imread(image, cv.IMREAD_GRAYSCALE)
    assert img is not None, "file could not be read, check with os.path.exists()"
    edges = cv.Canny(img,150,200)
    new_edge_path = EDGE_IMAGE_PATH + '/' + image.split('/')[1]
    cv.imwrite(new_edge_path,edges)
 
plt.subplot(121),plt.imshow(img,cmap = 'gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(edges,cmap = 'gray')
plt.title('Edge Image'), plt.xticks([]), plt.yticks([])
 
plt.show()
```
